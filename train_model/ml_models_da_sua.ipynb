# Dự báo nhiệt độ Hà Nội với Random Forest và XGBoost

Notebook chứa trực tiếp các bước xử lý và các lời gọi thư viện. Chạy lần lượt từ trên xuống.

- Input: **24 giờ** quan sát gần nhất, bước trượt **1 giờ**.
- Nhãn: `temperature_2m` **sau 1 giờ**.
- **Train đến hết 2024 → chọn cấu hình bằng validation 2025 → test trên dữ liệu 2026.**
- Chia tập theo thời điểm nhãn; giữ nguyên model fit trên train khi đánh giá test.



## 1. Import thư viện

Cell này chỉ import thư viện, bao gồm hai lớp model cần dùng. Tham số và các bước xử lý nằm ở những cell sau.

```python
from pathlib import Path
from time import perf_counter
import tempfile

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display
from numpy.lib.stride_tricks import sliding_window_view
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
```

## 2. Cấu hình bài toán và đường dẫn

Đổi `WINDOW_HOURS` tại đây nếu cần. Luồng chia tập hiện tại dùng `HORIZON_HOURS = 1` để dự báo sau 1 giờ. Tham số từng model nằm ngay trước cell train model đó. Mặc định dùng CSV processed đầy đủ đang có. Toàn bộ code xử lý, train và test nằm trong notebook này.


```python
PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "train_model":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / "data/processed/Hanoi_final.csv"
TEST_DATA_PATH = PROJECT_ROOT / "data/processed/Hanoi_test_final.csv"
OUTPUT_ROOT = PROJECT_ROOT / "artifacts/ml_sliding_window"
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
RUN_DIR = Path(tempfile.mkdtemp(prefix="run_", dir=OUTPUT_ROOT))

TARGET_COLUMN = "temperature_2m"
WINDOW_HOURS = 24
HORIZON_HOURS = 1
RANDOM_STATE = 42
N_JOBS = 4
VALIDATION_START = pd.Timestamp("2025-01-01")
```

## 3. Đọc dữ liệu

File processed đã có dữ liệu sạch và các cột sin/cos. Chỉ cần đọc CSV, chuyển `time` sang datetime và xem vài dòng đầu. Test 2026 sẽ được đọc riêng sau khi chốt model.

```python
data = pd.read_csv(DATA_PATH, parse_dates=["time"])
print("Số dòng, số cột:", data.shape)
display(data.head())
```

## 4. Chọn các biến đầu vào

Dùng trực tiếp 32 cột đã có sẵn trong file processed. Không dùng các cột thông tin địa điểm hoặc lag/rolling cũ. Các mô hình cây ở đây không cần scaler.

```python
WEATHER_FEATURES = [
    "temperature_2m", "relative_humidity_2m", "dew_point_2m",
    "apparent_temperature", "precipitation", "rain", "pressure_msl",
    "surface_pressure", "cloud_cover", "wind_speed_10m", "wind_speed_100m",
    "wind_direction_10m_sin", "wind_direction_10m_cos",
    "wind_direction_100m_sin", "wind_direction_100m_cos", "wind_gusts_10m",
    "et0_fao_evapotranspiration", "vapour_pressure_deficit",
    "soil_temperature_0_to_7cm", "soil_temperature_7_to_28cm",
    "soil_temperature_28_to_100cm", "soil_temperature_100_to_255cm",
    "soil_moisture_0_to_7cm", "soil_moisture_7_to_28cm",
    "soil_moisture_28_to_100cm", "soil_moisture_100_to_255cm",
    "weather_code", "is_day",
]
TIME_FEATURES = ["hour_sin", "hour_cos", "day_of_year_sin", "day_of_year_cos"]
INPUT_FEATURES = WEATHER_FEATURES + TIME_FEATURES
```

```python
input_columns = INPUT_FEATURES.copy()
```

## 5. Tạo cửa sổ trượt và chia train/validation

Mỗi bước được viết riêng: lấy dữ liệu theo giờ → tạo cửa sổ bằng `sliding_window_view()` → trải phẳng thành `X` → lấy nhiệt độ giờ tiếp theo làm `y` → chia tập bằng `train_test_split(..., shuffle=False)`.

Với `HORIZON_HOURS = 1`, `y = temperature[WINDOW_HOURS:]`: cửa sổ đầu dùng dòng 0–23 và đáp án ở dòng 24. Giờ mới nhất đứng trước trong mỗi hàng `X`, giữ đúng thứ tự input của model. Nhãn trước 01/01/2025 thuộc train; nhãn năm 2025 thuộc validation. Ví dụ 36 dòng theo giờ tạo 12 mẫu, mỗi mẫu có 24 × 32 = 768 cột.


```python
# 1. Lấy 32 feature và nhiệt độ theo từng giờ.
input_values = data[input_columns].to_numpy(dtype=np.float32)
temperature = data[TARGET_COLUMN].to_numpy(dtype=np.float32)
data_times = pd.DatetimeIndex(data["time"])

# 2. Tạo các cửa sổ 24 giờ; bỏ cửa sổ cuối vì chưa có đáp án sau 1 giờ.
input_windows = sliding_window_view(input_values, WINDOW_HOURS, axis=0)
input_windows = input_windows[:-1]
input_windows = input_windows.transpose(0, 2, 1)  # (số mẫu, số giờ, số feature)
input_windows = input_windows[:, ::-1, :]        # Giờ mới nhất đứng trước.

# 3. Mỗi hàng X chứa 24 × 32 giá trị; y là nhiệt độ ngay sau cửa sổ.
X = input_windows.copy().reshape(len(input_windows), -1)
y = temperature[WINDOW_HOURS:]
label_times = data_times[WINDOW_HOURS:]

# 4. Chia ngay theo năm của đáp án, giữ nguyên thứ tự thời gian.
train_count = int((label_times < VALIDATION_START).sum())
X_train, X_valid, y_train, y_valid = train_test_split(
    X, y,
    train_size=train_count,
    shuffle=False,
)

valid_label_times = label_times[train_count:]
```

```python
print("X_train:", X_train.shape, "| y_train:", y_train.shape)
print("X_valid:", X_valid.shape, "| y_valid:", y_valid.shape)
print(f"Mỗi mẫu: {WINDOW_HOURS} giờ × {len(input_columns)} feature = {X_train.shape[1]} cột.")
```

## 6. Chỉ số đánh giá và baseline

MAE/RMSE tính theo °C. Baseline persistence dự báo nhiệt độ tương lai bằng nhiệt độ tại giờ dự báo. Bảng validation và dự đoán của từng cấu hình sẽ được bổ sung ở các cell train bên dưới.

```python
# Giờ và nhiệt độ cuối mỗi cửa sổ, dùng để đánh giá validation.
forecast_times = data_times[WINDOW_HOURS - 1:-1]
current_temperature = temperature[WINDOW_HOURS - 1:-1]
valid_forecast_times = forecast_times[train_count:]
valid_baseline = current_temperature[train_count:]

def regression_metrics(actual, predicted):
    return {
        "MAE": float(mean_absolute_error(actual, predicted)),
        "RMSE": float(np.sqrt(mean_squared_error(actual, predicted))),
        "R2": float(r2_score(actual, predicted)),
    }

validation_scores = [{
    "model": "persistence", **regression_metrics(y_valid, valid_baseline), "fit_seconds": 0.0,
}]
validation_predictions = pd.DataFrame({
    "forecast_time": valid_forecast_times, "label_time": valid_label_times,
    "actual": y_valid, "persistence": valid_baseline,
})
trained_models = {}
display(pd.DataFrame(validation_scores))
```

## 7. Random Forest: khởi tạo rồi huấn luyện

Cell đầu gọi trực tiếp `RandomForestRegressor` từ scikit-learn. Cell tiếp theo gọi `.fit(X_train, y_train)` và `.predict(X_valid)`. Hai cấu hình chỉ khác độ sâu; mọi cấu hình đều fit trên cùng train và so sánh trên cùng validation.

```python
rf_models = {
    "rf_depth12": RandomForestRegressor(
        n_estimators=80, max_depth=12, min_samples_leaf=2, max_features="sqrt",
        random_state=RANDOM_STATE, n_jobs=N_JOBS,
    ),
    "rf_depth20": RandomForestRegressor(
        n_estimators=80, max_depth=20, min_samples_leaf=2, max_features="sqrt",
        random_state=RANDOM_STATE, n_jobs=N_JOBS,
    ),
}
rf_models
```

```python
# Có thể chạy lại cell này: thay kết quả RF cũ, không nhân đôi dòng trong bảng.
validation_scores = [row for row in validation_scores if row["model"] not in rf_models]
for name, model in rf_models.items():
    started = perf_counter()
    model.fit(X_train, y_train)
    fit_seconds = perf_counter() - started
    valid_prediction = model.predict(X_valid)
    score = regression_metrics(y_valid, valid_prediction)
    validation_scores.append({"model": name, **score, "fit_seconds": fit_seconds})
    validation_predictions[name] = valid_prediction
    trained_models[name] = model
    print(f"{name}: MAE={score['MAE']:.4f} °C; RMSE={score['RMSE']:.4f} °C; fit={fit_seconds:.1f}s")

display(pd.DataFrame([row for row in validation_scores if row["model"] in rf_models]))
```

## 8. XGBoost: khởi tạo rồi huấn luyện

Tương tự, gọi trực tiếp `XGBRegressor`. Không dùng test cho early stopping hoặc chọn tham số. Số cây được định trước; chọn giữa các cấu hình bằng validation 2025 ở bước sau.

```python
xgb_models = {
    "xgb_depth4": XGBRegressor(
        n_estimators=300, max_depth=4, learning_rate=0.05,
        objective="reg:squarederror", tree_method="hist", eval_metric="rmse",
        subsample=0.9, colsample_bytree=0.9, random_state=RANDOM_STATE, n_jobs=N_JOBS,
    ),
    "xgb_depth6": XGBRegressor(
        n_estimators=300, max_depth=6, learning_rate=0.05,
        objective="reg:squarederror", tree_method="hist", eval_metric="rmse",
        subsample=0.9, colsample_bytree=0.9, random_state=RANDOM_STATE, n_jobs=N_JOBS,
    ),
}
xgb_models
```

```python
validation_scores = [row for row in validation_scores if row["model"] not in xgb_models]
for name, model in xgb_models.items():
    started = perf_counter()
    model.fit(X_train, y_train)
    fit_seconds = perf_counter() - started
    valid_prediction = model.predict(X_valid)
    score = regression_metrics(y_valid, valid_prediction)
    validation_scores.append({"model": name, **score, "fit_seconds": fit_seconds})
    validation_predictions[name] = valid_prediction
    trained_models[name] = model
    print(f"{name}: MAE={score['MAE']:.4f} °C; RMSE={score['RMSE']:.4f} °C; fit={fit_seconds:.1f}s")

display(pd.DataFrame([row for row in validation_scores if row["model"] in xgb_models]))
```

## 9. Chọn một Random Forest và một XGBoost bằng validation

Chọn riêng cấu hình có MAE thấp nhất trong mỗi loại model; nếu bằng nhau thì dùng RMSE. Hai model được giữ trong `selected_models` để test trực tiếp.


```python
validation_metrics = pd.DataFrame(validation_scores)
model_ranking = validation_metrics.sort_values(["MAE", "RMSE"], kind="stable")

best_rf_name = model_ranking[model_ranking["model"].isin(rf_models)].iloc[0]["model"]
best_xgb_name = model_ranking[model_ranking["model"].isin(xgb_models)].iloc[0]["model"]
selected_models = {
    best_rf_name: trained_models[best_rf_name],
    best_xgb_name: trained_models[best_xgb_name],
}

validation_metrics["selected"] = validation_metrics["model"].isin(selected_models)
display(validation_metrics.sort_values(["MAE", "RMSE"]))
print("Hai model được chọn:", list(selected_models))
```

## 10. Đọc test và tạo X_test, y_test

Dùng trực tiếp hai model trong `selected_models` ở mục 9. Ghép 24 giờ cuối của `data` vào trước test rồi tạo cửa sổ giống mục 5.


```python
test_data = pd.read_csv(TEST_DATA_PATH, parse_dates=["time"])
test_data_with_history = pd.concat([data.tail(WINDOW_HOURS), test_data], ignore_index=True)

# Tạo cửa sổ như train, dự báo sau 1 giờ.
test_input_values = test_data_with_history[input_columns].to_numpy(dtype=np.float32)
test_input_windows = sliding_window_view(test_input_values, WINDOW_HOURS, axis=0)
test_input_windows = test_input_windows[:-1]
test_input_windows = test_input_windows.transpose(0, 2, 1)
test_input_windows = test_input_windows[:, ::-1, :]
X_test = test_input_windows.copy().reshape(len(test_input_windows), -1)

y_test = test_data[TARGET_COLUMN].to_numpy(dtype=np.float32)
test_label_times = pd.DatetimeIndex(test_data["time"])
print("X_test:", X_test.shape, "| y_test:", y_test.shape)
```

## 11. Test một RF và một XGBoost đã chọn

Dự báo trực tiếp bằng hai model đã chọn từ validation, tính MAE/RMSE/R² và so sánh với baseline persistence.


```python
test_temperature = test_data_with_history[TARGET_COLUMN].to_numpy(dtype=np.float32)
test_baseline = test_temperature[WINDOW_HOURS - 1:-1]
test_data_times = pd.DatetimeIndex(test_data_with_history["time"])
test_predictions = pd.DataFrame({
    "window_start": test_data_times[:len(y_test)],
    "forecast_time": test_data_times[WINDOW_HOURS - 1:-1],
    "label_time": test_label_times, "actual": y_test, "persistence": test_baseline,
})
test_scores = [{"model": "persistence", **regression_metrics(y_test, test_baseline)}]

for name, model in selected_models.items():
    prediction = model.predict(X_test)
    test_predictions[name] = prediction
    test_scores.append({"model": name, **regression_metrics(y_test, prediction)})

test_metrics = pd.DataFrame(test_scores)
test_metrics.to_csv(RUN_DIR / "test_metrics.csv", index=False)
test_predictions.to_csv(RUN_DIR / "test_predictions.csv", index=False)
validation_metrics.to_csv(RUN_DIR / "validation_metrics.csv", index=False)
validation_predictions.to_csv(RUN_DIR / "validation_predictions.csv", index=False)
display(test_metrics)
```

## 12. Thống kê theo tháng và biểu đồ


```python
monthly_rows = []
for month, part in test_predictions.groupby(test_predictions.label_time.dt.to_period("M")):
    for name in list(selected_models) + ["persistence"]:
        monthly_rows.append({
            "month": str(month), "model": name, "samples": len(part),
            **regression_metrics(part.actual, part[name]),
        })
monthly_metrics = pd.DataFrame(monthly_rows)
monthly_metrics.to_csv(RUN_DIR / "test_metrics_by_month.csv", index=False)
display(monthly_metrics)

first_week = test_predictions.iloc[:168]
fig, axes = plt.subplots(2, 1, figsize=(13, 8), constrained_layout=True)
axes[0].plot(first_week.label_time, first_week.actual, label="Observed", linewidth=1.5)
for name in selected_models:
    axes[0].plot(first_week.label_time, first_week[name], label=name, linewidth=1)
axes[0].set(title="First 7 days of test 2026", ylabel="Temperature (°C)")
axes[0].legend()
axes[0].grid(alpha=0.2)
for name, part in monthly_metrics.groupby("model"):
    axes[1].plot(part.month, part.MAE, marker="o", label=name)
axes[1].set(title="Monthly MAE on test 2026", ylabel="MAE (°C)", xlabel="Target month")
axes[1].legend()
axes[1].grid(alpha=0.2)
fig.savefig(RUN_DIR / "test_2026.png", dpi=150)
plt.show()
```
